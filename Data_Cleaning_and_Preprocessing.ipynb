# **Experiment No: 2 & 3**

**Title**

Importing, Exploring, Cleaning, Preprocessing, and Exporting a Sales Dataset using Python

**Aim**

To import a sales dataset from a CSV file, explore and clean the data, handle missing values, standardize text data, convert categorical variables into numerical form, normalize numerical features, and export the cleaned dataset using Python.


## **Dataset Used**

Sales Dataset (sales_data.csv) containing the following fields:

* OrderID

* Region

* Product

* Sales

* Quantity


---


##**Description of the Experiment**

In this experiment, a business-related sales dataset is analyzed using Python. The dataset is first imported and explored to understand its structure. Missing values and duplicate records are identified and handled. Text data is standardized to ensure consistency. Categorical variables such as region and product are converted into numerical form using one-hot encoding. Numerical features are normalized using StandardScaler. Finally, the cleaned dataset is exported to a new CSV file for further analysis.


---


##**Algorithm / Steps**

1. Import required Python libraries.

2. Load the CSV file using pd.read_csv().

3. Display the first few rows using head().

4. Display statistical summary using describe().

5. Check for missing values using isnull().

6. Remove duplicate records using drop_duplicates().

7. Handle missing values using fillna() and dropna().

8. Standardize text columns by removing extra spaces and converting to lowercase.

9. Convert categorical variables into numeric form using one-hot encoding.

10. Normalize numerical columns using StandardScaler.

11. Export the cleaned dataset to a new CSV file.


---


##**Input**

sales_data.csv – Sales dataset containing order details, region, product, sales amount, and quantity.

##**Output**

*   Display of first five rows of the dataset
*   Statistical summary of numerical columns
*   Missing value report
*   Cleaned and processed dataset saved as cleaned_sales_data.csv

---

##**Result**

The sales dataset was successfully imported, explored, cleaned, standardized, encoded, normalized, and exported. The cleaned dataset is now suitable for further business analysis and statistical modeling.

---

##**Conclusion**

This experiment demonstrates how Python can be effectively used for data cleaning and preprocessing in business applications. Proper handling of missing values, categorical data, and normalization improves the accuracy and reliability of data analysis in commerce-related domains.
